What is class in oop ?

```python
# English 

# A class is a blueprint or template used to create objects.

# Hindi

# Class ek blueprint/design hoti hai, jiske according hum objects create karte hain.
```

```python
#                   House ka Blueprint
#                         ↓
#                      Class
#                        ↓
#                  ┌─────┴─────┐
#                House 1    House 2
#                Object     Object

#  >>>>> Ek blueprint se multiple houses ban sakte hain.

# >>>>> Similarly, Python mein ek class se multiple objects bana sakte hain.
```

Class ka basic syntax

```python
class ClassName:
    # class ke andar code
    pass
```

Example

```python
class Student:
    pass


# class      → Python ka keyword
# Student    → Class ka naam
# pass       → Abhi class empty hai
```

class keyword kya karta hai?

```python
# class Python ko batata hai:
# Main ek naya class create kar raha hoon.

class Car:
    pass
```

Class ka naam kaise likhte hain?

```python
# Generally class ka naam PascalCase mein likhte hain.

class Student:
    pass

class BankAccount:
    pass

class MobilePhone:
    pass
```

Class ke andar kya hota hai?

```python
# Class ke andar hum generally attributes aur methods define karte hain.

# Example

class Student:

    name = "Suraj"
    age = 18

    def study(self):
        print("Student is studying")


# name → Attribute
# age  → Attribute
# study() → Method
```

Class alone kya karti hai?

Class ek design/template provide karti hai.

```python
class Student:
    name = "Suraj"
    age = 18

# Abhi humne sirf Student ka design banaya hai.

# Actual student object baad mein banega:

student1 = Student()
```

Ek class se multiple objects

```python
class Student:
    pass

student1 = Student()
student2 = Student()
student3 = Student()
```
