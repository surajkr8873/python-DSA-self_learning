What is object in python ?

```python
# English 

# An object is an instance of a class.

# Hindi 

# Object, class se banne wali actual entity hoti hai.
```

```python
# Class = Blueprint
# Object = Blueprint se bani actual cheez
```

Example

```python
# phle class banate hai

class student: # ye student ik class hai.
    
    pass

# uske baad hum object banate hai

s1 = student()  # ye s1 ush student ka class hai
```
