use recursion and print number from 1 to 10

```python
def number(n):
    if  n == 0:
        return 

    number(n - 1)
    print(n)

number(10)
```

sum of number from 1 to 10 use recursion

```python
def sum_number(n):
    if n == 0:
        return 0

    return (n) + sum_number(n - 1)

print("total_number =", sum_number(8))
```

Write  a program to findd the factorial of n. And use recursion

```python
def factorial(n):
    if n == 0:
        return 1

    fact= n * factorial(n - 1)
    # print(fact)
    return fact

factorial(5)
```
