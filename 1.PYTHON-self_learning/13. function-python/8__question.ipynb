Write a function to fiind the factorial of n (n is  the parameter)

```python
def fact_num(n):
    fact = 1
    for i in range(1 ,n+1):
        fact = fact * i
    
    print(fact)

fact_num(8)
```

write a function to print the largest number (number is the paraameter)

```python
def largest(a , b , c):
    if a > b and a > c:
        print("largest number", a)
        
    elif b > a and b > c:
        print("largest number" , b)

    else:
        print("largest number" , c)

largest(15 , 10 , 9)
    
```

Write a function to print the  number count

```python
def digit_count(n):
    count = 0

    while n > 0:
        n = n // 10
        count += 1

    print(count)

digit_count(123890)
```

Write a functiion to sum of all digit

```python
def sum_digit(n):
    digit = 0

    while n > 0:
        num = n % 10
        digit = digit + num
        n = n // 10

    print("total sum =",  digit)

sum_digit(78)
```

Write  a function to reverse a  number

```python
def reverse(n):
    digit = 0

    while n > 0:
        num = n % 10
        digit = digit * 10 + num
        n = n // 10

    print(digit)

reverse(1234)
```
