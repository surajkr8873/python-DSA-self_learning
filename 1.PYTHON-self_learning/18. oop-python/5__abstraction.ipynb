What is abstraction in oop

```python
# hiding the immplementation details of a class and only showing the essential features to the users

# Abstraction = unnecessary/internal details ko hide karna aur sirf important cheez user ko dikhana.
```

Simple language mein:

```python
# Andar kaise kaam ho raha hai, woh hide; kya kaam karna hai, woh show.
```

Example

```python
# Tum ATM use karte ho:

# Card insert kiya
# PIN dala
# Amount dala
# Cash mil gaya

# -------------------------------------------------------------------------------

# Tumhe ye nahi pata ki ATM ke andar:
# bank server se connection kaise hua
# account verify kaise hua
# transaction process kaise hua
# Ye sab hidden hai.

# ------------------------------------------------------------------------

# tumhe sirf neccesary chij mila
# pin >> amount >> cash
```

Python mein abstraction ke liye hum generally:

```python
from abc import ABC, abstractmethod
```

Chatgpt

```python
from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass

# Yahan Animal ek abstract class hai.
```

Har Animal ka sound() method hona chahiye, lekin Animal class ye decide nahi kar rahi ki sound kya hoga.

```python
class Dog(Animal):

    def sound(self):
        print("dog sound :", "Bark")


class Cat(Animal):

    def sound(self):
        print("cat sound :", "Meow")

# print("dog sound")

d = Dog()
d.sound()

# print("cat sound")

c = Cat()
c.sound()
```

```python
# Abstraction ka focus:
# KYA karna hai → show/define
# KAISE karna hai → child class par chhod dena
```
