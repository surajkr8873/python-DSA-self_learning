What is inheritance in oop ?

```python
# when one class (child) drives the property & methods of another class(parents).

# simple language inheritance.abs

# Inheritance = ek class ki properties aur methods ko doosri class me use kar pana.

# Animal → Parent class
# Dog → Child class

# Dog ek Animal hai, isliye Dog ko Animal ke features mil sakte hain.
```

EXAMPLE :-

```python
class Animal:            # isme animal jo hai wo parent class hai
    def eat(self):
        print("Animal is eating")


class Dog(Animal):       # isme animal ka child class dog hai
    pass


# eat() Dog class ke andar likha hi nahi hai, phir bhi Dog us method ko use kar sakta hai, kyunki woh Animal se inherit hua hai.
```

Example:


```python
# Parent Class
class Vehicle:
    def start(self):
        print("Vehicle started")

# Child Class
class Car(Vehicle):
    def drive(self):
        print("Car is driving")


# Object
c1 = Car()

c1.start()   # Parent class ka method
c1.drive()   # Child class ka method
```

Example 2 

```python
class Car:
    @staticmethod
    def start():
        print("car started")
    @staticmethod
    def stop():
        print("car stopped")

class Toyota_car(Car):
    def __init__(self, name):
        self.name = name

car1 = Toyota_car("fortuner")
car2 = Toyota_car("prius")

print(car1.start())
print(car1.name)
print(car2.name)
print(car1.name, "is ", car1.start())
```

Three  types of inheritance 

```python
# 1.single ineritance
# 2.multi level inheritance
# 3.multiple inheritance
```

Single inheritance

```python
# isme single parent class hoti hai or single derived class hoti hai.abs

# parent class 
# derived / child class
```

Example

```python
class Car:
    @staticmethod
    def start():
        print("car started")
    @staticmethod
    def stop():
        print("car stopped")

class Toyota_car(Car):
    def __init__(self, name):
        self.name = name

car1 = Toyota_car("fortuner")
car2 = Toyota_car("prius")

print(car1.start())
```

multi level inheritance 

```python
# isme ik base class hoti hai or ushse multiple derived class banti hai.

# parent class >> derived class >> derived class 
```

Example

```python
class car:
    @staticmethod
    def start():
        print("car started")

    @staticmethod
    def stop():
        print("car stopped")

class toyotacar(car):
    def __init__(self, brand):
        self.brand = brand

class fortuner(toyotacar):
    def __init__(self, type):
        self.type = type

car1 = car
car1 = toyotacar("toyota")
print(car1.brand)
car1.start()
```

Multiple inheritance

```python
# isme ik derived class hai to ye multiple parent class ko inherit kar skti hai.

# do ya ushse jada parent ik derived/child class ko inherit karne dete hai.
```

Example

```python
class A:
    varA = "welcome to class A"

class B:
    varB = "welcome to class B"

class C(A , B):
    varC = "welcome to class C"

C1 = C()

print(C1.varA)
print(C1.varB)
print(C1.varC)
```
