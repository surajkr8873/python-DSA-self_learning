What is oop in python ?

```python
# OOP = Object-Oriented Programming

# English :- OOP is a programming style where we organize our program using objects and classes.

# Hindi :- Simple language mein: real-world cheezon ko programming mein represent karna OOP hai.
```

Basic Exammple

```python
#                Real World
#                    ↓
#                    Car
#                     ↓
#             Color, Brand, Speed
#                     ↓
#           Start(), Stop(), Brake()


# >>> Properties → color, brand, speed
# >>> Actions → start, stop, brake
```

Python mein in cheezon ko hum Class aur Object ke through represent karte hain.

```python
# OOP ke 2 sabse basic concepts

#                  1. Class → Blueprint / design
#                  2. Object → Us blueprint se bani actual cheez
```

Example

```python
class Car:
    pass

car1 = Car()
car2 = Car()

# Car  → Class
# car1 → Object

# Class = Blueprint 
# Object = Blueprint se bani actual cheez
```
