INIT function in oop 

```python
#  CONSTRUCTOR

# All classes have a function called __ init __ , which is always executed when the class is being initiated. 
```

```python
# creating class

class Student:
    def __init__(self , name , age):
        self.name = name
        self.age = age

# creating object

s1 = Student("suraj kumar" , 17)
print(s1.name)
print(s1.age)
```

METHODS in oop 

```python
# Method :- Methods are functions that belongs to objects.
```

```python
# CREATING CLASS

class student:
    def __init__(self, name):
        self.name = name

    def hello(self):
        print("hello" , self.name)

#  CREATING OBJECT

s1 = student("suraj kumar")
s1.hello()
```

QUESTION:-

Create student class that takes name & marks oof three subjects as arguments in constructor. then create a method to print their average

```python
class student:
    def __init__(self ,name , marks):
        self.name = name
        self.marks = marks

    def get_avg(self):
        sum = 0
        for i in self.marks:
            sum += i
        print("name :",self.name, "and", "avg_marks :",  sum/3)

s1 = student("surja kumar" , [98 , 96 , 98])
s1.get_avg()
        
```

Static method

```python
# Method that don't use the self parameter (work at class level)
```

```python
class student:
    @staticmethod           # decorator
    def college():
        print("abc college")

s1 = student()
s1.college()
```

What is decorator

```python
# Decorator allows us to wrap another function in order to extend the behaviour of the wrapped function,without permanently modifying it
```
