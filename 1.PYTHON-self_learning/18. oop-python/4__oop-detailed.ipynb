Class method

```python
# A class method is bound to the class & recieves the class ass an implicit first argument.

#  @classmethod aisa method hota hai jo object ke bajay class ko access karta hai.

# Isme first parameter normally cls hota hai.
```

Basic Example

```python
class Student:
    school = "ABC School"

    @classmethod
    def show_school(cls):
        print(cls.school)


Student.show_school()

print("===================== Example no 2 ==========================================================")

class person:
    name = "anonymous"

    @classmethod
    def changename(cls, name):
        cls.name = name

p1 = person()
p1.changename("rahul kumar")
print(p1.name)
print(person.name)

```

Property method

```python
# we use @property decorator on any method in the class to use the  method as a prooperty.

# @property ka use method ko attribute/property ki tarah access karne ke liye hota hai.
```

Example

```python
class Student:
    def __init__(self, name):
        self._name = name

    def get_name(self):
        return self._name


s1 = Student("Suraj kumar")

print(s1.get_name())

print("==================== Exammpe no 2 =======================================")

class student:
    def __init__(self, che, phy, math):
        self.che = che
        self.phy = phy
        self.math = math

    @property
    def percentage(self):
        return str((self.phy + self.che + self.math) / 3) + "%"

student1 = student(98, 97, 78)
print(student1.percentage)

student1.phy = 86
print(student1.percentage)
```

```python
# Q1

# Define a circle class to create a circle with radius r usingg the coonstructor.define an are() method of the class which claculate the area of the circle.define a perimeter() method  of the class which allows  you to calculate the perimeter of this circle.

class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 2 * (22/7) * self.radius ** 2

    def perimeter(self):
        return 2 * (22/7) * self.radius

C1 = Circle(7)
print(C1.area())
print(C1.perimeter())
```

```python
# Q2.

# definr a employe class with attribute role,department & salary.This class also has show details() method. create an engineer class that inherit properties from employe & has additional attribute;anme & age.

class employee:
    def __init__(self, role , dept, salary):
        self.role = role
        self.dept = dept
        self.salary = salary

    def showdetails(self):
        print("role :", self.role)
        print("dept :", self.dept)
        print("salary :", self.salary)

class engineer(employee):
    def __init__(self, name, age):
        self.name = name
        self.age = age
        super().__init__("engineer", "it" , "75,000")

eng1 = engineer("suraj kumar,", 40)
print("name :", eng1.name, "age :" ,eng1.age)
eng1.showdetails()
```

```python
# Q3.

# Create a class called order which stores item & its price.use  dunder function__gt__ to convey that: order1 > order2 if price of order1 > price of  order2

class Order:
    def __init__(self, item, price):
        self.item = item
        self.price = price

    def __gt__(self, ord2):
        return self.price > ord2.price

ord1 = Order("kurkure", 20)
ord2 = Order("tea", 10)

print(ord1 > ord2)
```
