```python

from abc import ABC, abstractmethod

# Abstraction
class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass


# Encapsulation + Inheritance
class Bankaccount(Payment):
    def __init__(self, name, balance):
        self.name = name
        self.__balance = balance

    def pay(self, amount):
        if amount <= 0:
            print("Invalid payment amount")
        elif amount > self.__balance:
            print("Insufficient balance")
        else:
            self.__balance -= amount
            print("Payment successful:", amount)

    def deposit(self, amount):
        if amount <= 0:
            print("Invalid deposit amount")
        else:
            self.__balance += amount
            print("Deposit successful:", amount)

    def get_balance(self):
        return self.__balance


# UPI payment
class UPI(Payment):
    def pay(self, amount):
        print("UPI payment:", amount)


# Card payment
class Card(Payment):
    def pay(self, amount):
        print("Card payment:", amount)


# Objects
bank = Bankaccount("Suraj Kumar", 1000)
upi = UPI()
card = Card()

# Bank account se payment
bank.pay(300)
print("Balance:", bank.get_balance())

# Deposit
bank.deposit(500)
print("Balance:", bank.get_balance())

# Other payment methods
upi.pay(200)
card.pay(400)
```
