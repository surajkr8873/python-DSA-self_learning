```python

class Bankaccount:
    def __init__(self, name, balance):
        self.name = name

        if balance < 0:
            raise ValueError("Balance cannot be negative")

        self.__balance = balance

    def get_balance(self):
        return self.__balance

    # Deposit
    def deposit(self, amount):
        if amount <= 0:
            print("Deposit amount must be positive")
        else:
            self.__balance += amount
            print("Deposit successful. New balance:", self.__balance)

    # Withdraw
    def withdraw(self, amount):
        if amount <= 0:
            print("Withdraw amount must be positive")
        elif amount > self.__balance:
            print("Insufficient balance")
        else:
            self.__balance -= amount
            print("Withdraw successful. New balance:", self.__balance)


account1 = Bankaccount("suraj kumar", 100)

account1.deposit(300)
account1.withdraw(250)
account1.withdraw(50)
account1.deposit(30)
account1.withdraw(500)

print("Final balance:", account1.get_balance())
```

```python
class Profile:
    def __init__(self, username, password):
        self.username = username
        self.__password = password

    # setter
    def set_new_password(self, new_password):
        self.__password = new_password
        print("password update")

p1 = Profile("ethereal_suraj","suraj8873")
print(p1.username)
p1.set_new_password(123456)
```

property method

```python
class Student:
    def __init__(self, name, roll_no, marks):
        self.name = name
        self.roll_no = roll_no
        self.__marks = marks

    @property                  # ye method se hum kisi function ko as a variable to dekh kste hai lekin hum change nahi kar skte pvt chijo ko
    def marks(self):
        return self.__marks

    @marks.setter
    def marks(self, new_marks):
        if not 0 <= new_marks <= 100:
            print("Invalid marks")
        else:
            self.__marks = new_marks

s1 = Student("suraj kumar", 6, 52)
print(s1.name)
s1.marks = 99
print(s1.marks)
```
