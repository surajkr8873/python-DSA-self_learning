What is encapsulation in oop ?

```python
# Wrapping data and function into a single unit(object)

# Encapsulation = data aur us data par kaam karne wale methods ko ek hi class ke andar bind/pack karna, aur data ko direct access se protect karna.
```

```python
# Data ko class ke andar safe rakhna aur usko controlled way se access karna = Encapsulation.
```

```python
# Socho tumhare paas Bank Account hai.
# Account mein:
# balance = rs.50,000
# Agar koi bhi directly balance change kar de:
# balance = rs.1
# to problem ho jayegi

# Isliye balance ko directly access/change karne ke bajay methods ke through control karte hain:

# deposit()
# withdraw()
# get_balance()

# Yahi idea Encapsulation hai.
```

Python mein Encapsulation

Python mein commonly _ aur __ ka use karke attributes ko protect/private-like banaya jata hai.

```python
class BankAccount:

    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount

    def get_balance(self):
        return self.__balance

account = BankAccount(50000)

account.deposit(5000)

print(account.get_balance())
```

Encapsulation ka structure

```python
        #                  CLASS
        #           ┌─────────────────┐
        #           │                 │
        #           │  __balance      │  ← DATA
        #           │                 │
        #           │  deposit()      │  ← METHOD
        #           │  withdraw()     │  ← METHOD
        #           │  get_balance()  │  ← METHOD
        #           │                 │
        #           └─────────────────┘
        #                   ↓
        #            Controlled Access
```

Ek line mein yaad karo:

```python
# Abstraction → implementation hide
# Encapsulation → data protect + controlled access
```
