What is recursion in python ?

```python
# English 

# Recursion is a process in which a function calls itself again and again until a stopping condition is reached.

# Hindi 

# Recursion mein ek function khud ko hi call karta hai, aur ye process tab tak chalta hai jab tak stopping condition (base case) nahi aa jaati.
```

Simple structure  

```python
def function():
    # kuch kaam

    function()   # function khud ko call kar raha hai
```

1. Base Case    → Recursion ko rokta hai 
2. Recursive Call → Function khud ko call karta hai 

Basiic example

```python
def count(n):

    if n == 0:       # Base Case
        return

    print(n)

    count(n - 1)     # Recursive Call


count(5)
```

```python
# Function khud ko call kare = Recursion
# Recursion ko rokne wali condition = Base Case
# Khud ko dobara call karna = Recursive Call
```

DEPTH OF RECURSION

Normal function

```python
def hello():
    print("Hello")

hello()

# hello() → function chala
#         → Hello print hua
#         → function khatam 
```

Ab recursion mein function khud ko hi call karta hai:

```python
def hello():
    print("Hello")
    hello()


# yaha recursioon rukega hi nahi.
# isiliye recursion me stopinng condition li jati hai.
```

Exammple

```python
def count(n):

    if n == 0:
        return

    print(n)
    count(n - 1)

count(5)
```

```python
def number(n):
    if n == 0:
        return
    
    print(n)
    number(n-1)

number(8)
```

```python
def print_numbers(n):

    if n == 0:          # Base Case
        return

    print_numbers(n - 1) # Recursive Call
    print(n)


print_numbers(5)
```

1 se n number tak ka  sum

```python
def sum_numbers(n):

    if n == 0:          # Base Case
        return 0

    return n + sum_numbers(n - 1)


result = sum_numbers(5)

print(result)
```
